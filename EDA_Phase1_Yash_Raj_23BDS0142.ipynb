# 5. Handling Missing Data

Missing values can affect the quality and reliability of data analysis. Therefore, it is important to identify missing values, determine the number of missing records in each column, and apply suitable techniques to handle them before proceeding with further analysis.

```python
print("Missing Values in Each Column:")

df.isnull().sum()
```

## Observation

- The dataset contains **missing values in several features**, while the **Result** column has no missing values.
- The **Gender of the patient** column has the highest number of missing values (**902**), followed by **Alkphos Alkaline Phosphotase (796)** and **Total Bilirubin (648)**.
- Other clinical variables such as **Direct Bilirubin, SGPT, SGOT, Total Proteins, Albumin, and A/G Ratio** also contain missing values, ranging from **462 to 561** records.
- The **Age of the patient** column has only **2 missing values**, indicating that it is nearly complete.
- Since missing values are present in multiple important clinical variables, **appropriate imputation techniques (such as mean, median, or mode imputation) should be applied** before performing further analysis or building machine learning models.
- Proper handling of missing data helps improve the quality, reliability, and predictive performance of the dataset.

```python
numerical_columns = df.select_dtypes(include=['float64', 'int64']).columns

for column in numerical_columns:
    if column != 'Result':
        df[column] = df[column].fillna(df[column].median())

df['Gender of the patient'] = df['Gender of the patient'].fillna(
    df['Gender of the patient'].mode()[0]
)

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
sns.pairplot(
    df[
        [
            "Age of the patient",
            "Total Bilirubin",
            "Direct Bilirubin",
            "Total Protiens"
        ]
    ]
)

plt.show()
```

## 10.2 Correlation Matrix

A correlation matrix is used to measure the strength and direction of the linear relationship among numerical variables. The correlation coefficient ranges from **-1 to +1**, where values closer to **+1** indicate a strong positive correlation, values closer to **-1** indicate a strong negative correlation, and values near **0** indicate little or no correlation.

```python
correlation = df.corr(numeric_only=True)

correlation
```

## 10.3 Correlation Heatmap

A heatmap provides a visual representation of the correlation matrix. Different colors represent different correlation strengths, making it easier to identify highly correlated variables within the dataset.

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Liver Disease Dataset")

plt.show()
```

### 2.1.2 Measures of Central Tendency

Measures of central tendency provide a representative value for a set of observations. Different measures are used because the clinical variables may have different distributions and may contain extreme observations.

### 2.1.2.1 Calculation of Central Tendency Measures

The central tendency measures are calculated for each numerical clinical variable to obtain different representations of its typical value.

The arithmetic mean, median, mode, geometric mean, harmonic mean, and quadratic mean are calculated and compared for the selected clinical variables.

### 2.1.3 Weighted Measures of Central Tendency

Weighted measures of central tendency assign different importance to observations using a set of weights. They are useful for demonstrating how the representative value changes when observations are given different weights.

In this analysis, weighted arithmetic mean, weighted geometric mean, and weighted harmonic mean are calculated for the numerical clinical variables using sequential weights.

The weighted results are compared with the corresponding unweighted measures to observe the effect of applying weights.

```python
from scipy.stats import gmean, hmean

weighted_results = []

for col in numeric_features:
    data = df[col].dropna().to_numpy()
    weights = np.arange(1, len(data) + 1)

    weighted_arithmetic = np.average(data, weights=weights)
    weighted_geometric = np.exp(
        np.average(np.log(data), weights=weights)
    )
    weighted_harmonic = weights.sum() / np.sum(weights / data)

    weighted_results.append({
        "Variable": col,
        "Weighted Arithmetic Mean": weighted_arithmetic,
        "Weighted Geometric Mean": weighted_geometric,
        "Weighted Harmonic Mean": weighted_harmonic
    })

weighted_results_df = pd.DataFrame(weighted_results)

weighted_results_df
```

### Observation

The weighted measures show how the calculated central values change when different observations are assigned different weights.

- The **weighted arithmetic means** are close to the corresponding ordinary arithmetic means for most variables.
- **Total Bilirubin, Direct Bilirubin, SGPT, and SGOT** have weighted geometric and harmonic means that remain much lower than their arithmetic means, reflecting the effect of their higher-valued observations.
- **Alkphos, SGPT, and SGOT** show larger differences between the weighted arithmetic, geometric, and harmonic means.
- **Total Proteins, Albumin, and A/G Ratio** show relatively small changes across the weighted measures.
- The weighted results demonstrate that the choice of weighting and type of mean can produce different representations of the same clinical variable.

### 2.1.4 Alpha Means

The alpha mean is a generalized measure of central tendency in which the value of the mean depends on the selected value of α.

For this analysis, α = -1, 1, and 2 are considered. These values correspond to the harmonic mean, arithmetic mean, and quadratic mean respectively.

The alpha means are calculated for all nine numerical clinical variables to compare how the representative value changes with different values of α.

```python
alpha_values = [-1, 1, 2]

alpha_results = []

for col in numeric_features:
    data = df[col].dropna().to_numpy()

    row = {"Variable": col}

    for alpha in alpha_values:
        if alpha == 0:
            value = np.exp(np.mean(np.log(data)))
        else:
            value = np.mean(data ** alpha) ** (1 / alpha)

        row[f"Alpha Mean (α={alpha})"] = value

    alpha_results.append(row)

alpha_means_df = pd.DataFrame(alpha_results)

alpha_means_df
```

### Observation

The alpha means show how the measure of central tendency changes as the value of α changes.

- For all variables, the **α = -1 mean** is lower than the **α = 1 mean**, while the **α = 2 mean** is higher.
- For **Age**, the values increase from **34.88** at α = -1 to **43.81** at α = 1 and **46.81** at α = 2.
- **Total Bilirubin, Direct Bilirubin, SGPT, and SGOT** show particularly large increases as α changes from -1 to 2, indicating that higher observations have a strong effect on the quadratic mean.
- **Total Proteins, Albumin, and A/G Ratio** show comparatively smaller changes across the three alpha values.
- The results demonstrate that increasing α gives greater influence to larger observations, while negative α values give greater influence to smaller observations.

###Observation

### 2.1.6.6 Mean Deviation from Mean

Mean deviation from the mean measures the **average absolute distance of observations from the arithmetic mean**.

**Formula:**

**Mean Deviation from Mean = Σ|xᵢ − x̄| / n**

where:

- **xᵢ** represents an individual observation.

- **xᵢ** represents an individual observation.
- **M** represents the median.
- **n** represents the total number of observations.

This measure is useful for describing variability around the median and is less affected by extreme values than measures based on the mean.

### Variance

Variance measures the degree of variability of observations around their arithmetic mean.

It is calculated as the average of the squared deviations of the observations from the mean.

**Formula:**

Variance = Σ(xᵢ − x̄)² / n

where:

- xᵢ = individual observation
- x̄ = arithmetic mean
- n = number of observations

A higher variance indicates greater variability in the observations.

```python
variance_results = []

for col in numeric_features:
    data = df[col].dropna()

    variance_results.append({
        "Variable": col,
        "Variance": data.var(ddof=0)
    })

variance_df = pd.DataFrame(variance_results)

variance_df
```

- **SGOT** has the highest variance of approximately **75,783.26**, indicating very high variability.
- **Alkphos** has a variance of approximately **55,775.61**, while **SGPT** has a variance of approximately **32,218.10**.

- **Age** has a variance of approximately **271.55**.
- **Total Proteins, Albumin, and A/G Ratio** have comparatively smaller variances of **1.16**, **0.62**, and **0.10**, respectively.
- Overall, the variance results are consistent with the standard deviation and other dispersion measures, showing that **SGOT, Alkphos, and SGPT have substantially greater variability** than the other clinical variables.

### 2.1.7.5 Frequency Polygon

A frequency polygon is a graphical representation of a frequency distribution in which the **midpoints of the class intervals** are plotted against their corresponding frequencies and connected using straight lines.

For this analysis, a frequency polygon is constructed for **Total Bilirubin** using its class midpoints and frequencies.

For Total Bilirubin, the frequency values are smoothed using interpolation to obtain a continuous curve.

### 2.1.7.7 Ogive

An ogive is a **cumulative frequency curve** that shows the cumulative number of observations up to or above a particular value.

There are two common types:

- **Less-than ogive:** shows the cumulative frequency up to each upper class boundary.
- **More-than ogive:** shows the cumulative frequency from each lower class boundary onward.

For this analysis, a **less-than ogive** is constructed for Total Bilirubin using the upper class boundaries and cumulative frequencies.

```python
bilirubin_bins = np.linspace(
    df["Total Bilirubin"].min(),
    df["Total Bilirubin"].max(),
    11
)

bilirubin_frequency = pd.cut(
    df["Total Bilirubin"],
    bins=bilirubin_bins,
    include_lowest=True
).value_counts().sort_index()

cumulative_frequency = bilirubin_frequency.cumsum()

upper_boundaries = [
    interval.right for interval in bilirubin_frequency.index
]

plt.figure(figsize=(10, 6))

plt.plot(
    upper_boundaries,
    cumulative_frequency.values,
    marker="o"
)

plt.title("Less-than Ogive of Total Bilirubin")
plt.xlabel("Upper Class Boundary")
plt.ylabel("Cumulative Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Observation

The less-than ogive shows the cumulative frequency of **Total Bilirubin** as the upper class boundary increases.

- The cumulative frequency increases rapidly in the lower Total Bilirubin ranges because most observations are concentrated at lower values.
- By an upper boundary of approximately **8**, the cumulative frequency has already reached around **17,200 observations**.
- The curve continues to rise up to approximately **30**, reaching about **18,900 observations**.
- After approximately **40**, the curve becomes almost flat, indicating that only a small number of additional observations occur at higher Total Bilirubin values.
- The curve reaches approximately **19,083 observations** at the highest boundary, representing the total number of observations.
- Overall, the ogive confirms that **most Total Bilirubin observations occur at relatively low values**, with comparatively few observations at higher values.

### 2.1.7.8 Smoothed Ogive

A smoothed ogive is a smooth representation of a cumulative frequency distribution.

It reduces the sharp changes between class boundaries and provides a continuous view of how the cumulative frequency increases across the distribution.

For Total Bilirubin, the cumulative frequencies are smoothed using interpolation to visualize the overall cumulative distribution.

```python
from scipy.interpolate import make_interp_spline

x = np.array(upper_boundaries)
y = np.array(cumulative_frequency.values)

x_smooth = np.linspace(x.min(), x.max(), 300)

spline = make_interp_spline(x, y, k=3)
y_smooth = spline(x_smooth)

y_smooth = np.clip(
    y_smooth,
    cumulative_frequency.min(),
    cumulative_frequency.max()
)

plt.figure(figsize=(10, 6))

plt.plot(
    x_smooth,
    y_smooth
)

plt.title("Smoothed Ogive of Total Bilirubin")
plt.xlabel("Upper Class Boundary")
plt.ylabel("Cumulative Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Observation

The smoothed ogive shows the cumulative frequency distribution of **Total Bilirubin**.

- The curve becomes almost flat after approximately **40**, indicating that relatively few additional observations occur at higher Total Bilirubin values.
- The curve approaches the total number of observations, approximately **19,083**, at the highest class boundary.
- Overall, the smoothed ogive confirms that **most observations have relatively low Total Bilirubin values**, while only a small proportion occurs at higher values.

Skewness measures the asymmetry of a probability distribution around its mean.

### 2.2.1.6 Expected Frequencies

Expected frequencies represent the number of observations expected in each cell of the contingency table if the two categorical variables are statistically independent.

The expected frequency for each cell is calculated using the row total, column total, and grand total.

**Formula:**

Expected Frequency = (Row Total × Column Total) / Grand Total

The observed and expected frequencies are later compared using the Chi-Square test of independence.

```python
expected_frequencies = pd.DataFrame(
    np.outer(
        contingency_table.sum(axis=1),
        contingency_table.sum(axis=0)
    ) / contingency_table.values.sum(),
    index=contingency_table.index,
    columns=contingency_table.columns
)

expected_frequencies
```

### Observation

The expected frequencies represent the cell counts that would be expected if Gender and Result were independent.

- For **Gender 0, Result 1**, the expected frequency is approximately **3955.12**, compared with the observed frequency of **3957**.
- For **Gender 0, Result 2**, the expected frequency is approximately **1590.88**, compared with the observed frequency of **1589**.
- For **Gender 1, Result 1**, the expected frequency is approximately **9653.88**, compared with the observed frequency of **9652**.
- For **Gender 1, Result 2**, the expected frequency is approximately **3883.12**, compared with the observed frequency of **3885**.
- The observed and expected frequencies are very close to each other in all four cells.
- The expected frequencies provide the basis for calculating the **Chi-Square contributions** and performing the **Chi-Square test of independence**.

### 2.2.1.7 Chi-Square Contribution

The Chi-Square contribution measures how much each cell contributes to the overall Chi-Square statistic.

It compares the observed frequency with the expected frequency for each cell.

**Formula:**

Chi-Square Contribution = (Observed − Expected)² / Expected

The contributions from all cells are summed to obtain the overall Chi-Square statistic.

```python
chi_square_contributions = (
    (contingency_table - expected_frequencies) ** 2
) / expected_frequencies

chi_square_contributions
```

### Observation

The Chi-Square contributions show the contribution of each cell to the overall Chi-Square statistic.

- The contribution for **Gender 0, Result 1** is approximately **0.000896**.
- The contribution for **Gender 0, Result 2** is approximately **0.002227**.
- The contribution for **Gender 1, Result 1** is approximately **0.000367**.
- The contribution for **Gender 1, Result 2** is approximately **0.000912**.
- All four contributions are very small because the observed frequencies are very close to their corresponding expected frequencies.
- The sum of these four contributions gives the overall **Chi-Square statistic**.

### 2.2.1.8 Pearson Chi-Square Test of Independence

The Pearson Chi-Square test is used to determine whether there is a statistically significant association between two categorical variables.

For this analysis, the test examines whether **Gender** and **Result** are independent.

**Hypotheses:**

- **H₀:** Gender and Result are independent.
- **H₁:** Gender and Result are associated.

**Formula:**

Chi-Square = Σ[(Observed − Expected)² / Expected]

The degrees of freedom are calculated as:

**df = (r − 1)(c − 1)**

where r is the number of rows and c is the number of columns.

```python
from scipy.stats import chi2_contingency

chi2_stat, p_value, degrees_of_freedom, expected = chi2_contingency(
    contingency_table
)

print("Chi-Square Statistic:", chi2_stat)
print("p-value:", p_value)
print("Degrees of Freedom:", degrees_of_freedom)
```

### Observation

The Pearson Chi-Square test was used to examine the association between Gender and Result.

- The calculated **Chi-Square statistic is 0.00237**.
- The **degrees of freedom are 1**.
- The **p-value is 0.96114**.
- Using a significance level of **0.05**, the p-value is greater than 0.05.
- Therefore, the null hypothesis of independence is **not rejected**.
- Based on this test, the data do not provide statistically significant evidence of an association between Gender and Result.

### 2.2.1.9 Phi Coefficient

The Phi coefficient measures the strength of association between two categorical variables in a 2 × 2 contingency table.

Its value ranges from **−1 to +1**, where values closer to zero indicate a weaker association.

For a 2 × 2 table, the Phi coefficient can be calculated as:

**Phi = √(Chi-Square / n)**

where:

- Chi-Square = Pearson Chi-Square statistic
- n = total number of observations

```python
n = contingency_table.values.sum()

phi_coefficient = np.sqrt(chi2_stat / n)

print("Phi Coefficient:", phi_coefficient)
```

### Observation

The Phi coefficient measures the strength of association between Gender and Result.

- The calculated **Phi coefficient is 0.000353**.
- The value is extremely close to **0**, indicating a negligible association between the two categorical variables.
- This is consistent with the Pearson Chi-Square test, which produced a very large p-value of **0.96114**.
- Therefore, the Phi coefficient and Chi-Square test provide consistent results for the Gender × Result relationship.

### 2.2.1.10 Cramer's V

Cramer's V measures the strength of association between two categorical variables.

Its value ranges from **0 to 1**, where a value close to 0 indicates a very weak association and a value closer to 1 indicates a stronger association.

For a contingency table, Cramer's V is calculated as:

**Cramer's V = √(Chi-Square / (n × min(r − 1, c − 1)))**

where:

- Chi-Square = Pearson Chi-Square statistic
- n = total number of observations
- r = number of rows
- c = number of columns

```python
r, c = contingency_table.shape

cramers_v = np.sqrt(
    chi2_stat / (n * min(r - 1, c - 1))
)

print("Cramer's V:", cramers_v)
```

### Observation

Cramer's V measures the strength of association between Gender and Result.

- The calculated **Cramer's V is 0.000353**.
- The value is extremely close to **0**, indicating a negligible association between the two categorical variables.
- For this 2 × 2 contingency table, Cramer's V has the same numerical value as the Phi coefficient.
- The result is consistent with the Pearson Chi-Square test, which produced a p-value of **0.96114**.

### 2.2.1.11 Tschuprow's T

Tschuprow's T is another measure of association between two categorical variables.

It is based on the Chi-Square statistic and takes into account the number of rows and columns in the contingency table.

**Formula:**

T = √[Chi-Square / √((r − 1)(c − 1)) × n]

where:

- Chi-Square = Pearson Chi-Square statistic
- n = total number of observations
- r = number of rows
- c = number of columns

```python
tschuprow_t = np.sqrt(
    chi2_stat / (
        n * np.sqrt((r - 1) * (c - 1))
    )
)

print("Tschuprow's T:", tschuprow_t)
```

### Observation

Tschuprow's T measures the strength of association between Gender and Result.

- The calculated **Tschuprow's T is 0.000353**.
- The value is extremely close to **0**, indicating a negligible association between the two categorical variables.
- The result is consistent with the **Phi coefficient** and **Cramer's V**, which also have a value of approximately **0.000353**.
- The association measures are consistent with the Pearson Chi-Square test, which produced a **p-value of 0.96114**.

# 2.4 K-Means Clustering
### 2.4.1 Preparation of Data for K-Means

K-Means clustering groups observations based on the similarity of their numerical characteristics.

### 2.4.2 Elbow Method

The Elbow Method is used to determine a suitable number of clusters for K-Means.

K-Means is applied for different values of K, and the **Within-Cluster Sum of Squares (WCSS)** is calculated for each value.

As K increases, WCSS generally decreases. The point where the decrease begins to slow down forms the **elbow**, which provides a useful indication of an appropriate number of clusters.

```python
from sklearn.preprocessing import StandardScaler

X = df[numeric_features].copy()

print("Missing values:")
print(X.isna().sum())

X = X.fillna(X.median())

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=numeric_features,
    index=df.index
)

print("\nNaN values after preprocessing:")
print(X_scaled.isna().sum())

from sklearn.cluster import KMeans

wcss = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), wcss, marker="o")
plt.title("Elbow Method for K-Means Clustering")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.xticks(range(2, 11))
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Observation

The Elbow Method shows a continuous decrease in WCSS as the number of clusters increases.

- WCSS decreases sharply from **K = 2 to K = 4**.
- After **K = 4**, the reduction in WCSS becomes more gradual.
- The curve begins to flatten after K = 4, indicating diminishing improvement from adding additional clusters.
- Therefore, **K = 4 is a suitable candidate** based on the Elbow Method.
- The final number of clusters will be further evaluated using the **Silhouette Score**.

### 2.4.3 Silhouette Analysis

The Silhouette Score is used to evaluate the quality of clustering for different values of K.

It measures how well each observation fits within its assigned cluster compared with the nearest other cluster.

The Silhouette Score ranges from **−1 to +1**.

- A value close to **+1** indicates well-separated clusters.
- A value close to **0** indicates overlapping clusters.
- A negative value indicates that observations may have been assigned to an inappropriate cluster.

The Silhouette Score is calculated for different values of K to support the selection of the final number of clusters.

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores.append(score)

plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), silhouette_scores, marker="o")
plt.title("Silhouette Score for Different Numbers of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.xticks(range(2, 11))
plt.grid(True)
plt.tight_layout()
plt.show()
```

### Observation

The Silhouette analysis evaluates the quality of clustering for different values of K.

- The highest Silhouette Score is obtained at **K = 2**, with a score of approximately **0.496**.
- The score decreases sharply when moving from K = 2 to K = 3.
- The scores for K = 4 and K = 5 are approximately **0.246** and **0.251**, respectively.
- For K ≥ 6, the Silhouette Score remains around 0.18–0.20.
- The Elbow Method suggested **K = 4** as a candidate, while the Silhouette Method shows stronger cluster separation at **K = 2**.
- Since the Silhouette Score is substantially higher for K = 2, **K = 2 will be used for the final K-Means clustering**.

### 2.4.4 Final K-Means Clustering

Based on the Elbow Method and Silhouette Analysis, K = 2 is selected for the final K-Means model.

The K-Means algorithm is now applied with two clusters using the standardized numerical clinical variables.

Each observation is assigned to the cluster whose centroid is closest according to Euclidean distance.

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_scaled)

df["KMeans_Cluster"] = kmeans_labels

df["KMeans_Cluster"].value_counts().sort_index()
```

### 2.4.6 Euclidean Distance Calculation

K-Means assigns each observation to the cluster whose centroid is closest to the observation.

The Euclidean distance between an observation and a centroid is calculated as:

Distance = √[Σ(xᵢ − cᵢ)²]

where:

- xᵢ = value of the observation for feature i
- cᵢ = corresponding centroid value
- i = each numerical feature

The calculation is performed using the standardized numerical features so that differences in the original measurement scales do not dominate the distance.

```python
from scipy.spatial.distance import cdist

centroids = kmeans.cluster_centers_

distances = cdist(
    X_scaled,
    centroids,
    metric="euclidean"
)

distance_table = pd.DataFrame(
    distances,
    columns=["Distance to Cluster 0", "Distance to Cluster 1"],
    index=df.index
)

distance_table.head()
```

### Observation

The Euclidean distance calculation shows how each observation is assigned to the nearest K-Means centroid.

- **Observation 0:** Distance to Cluster 0 = **1.433916**, Distance to Cluster 1 = **4.959782** → assigned to **Cluster 0**.
- **Observation 1:** Distance to Cluster 0 = **3.321687**, Distance to Cluster 1 = **3.153043** → assigned to **Cluster 1**.
- **Observation 2:** Distance to Cluster 0 = **2.157324**, Distance to Cluster 1 = **3.363823** → assigned to **Cluster 0**.
- **Observation 3:** Distance to Cluster 0 = **1.065603**, Distance to Cluster 1 = **4.843367** → assigned to **Cluster 0**.
- **Observation 4:** Distance to Cluster 0 = **2.842337**, Distance to Cluster 1 = **4.537111** → assigned to **Cluster 0**.

Thus, K-Means assigns each observation to the cluster having the **smaller Euclidean distance** from its centroid.

**Formula:**

Distance = √[Σ(xᵢ − yᵢ)²]

where:

- xᵢ = standardized value of feature i for the first observation
- yᵢ = standardized value of feature i for the second observation
- i = each numerical clinical feature

A smaller Euclidean distance indicates greater similarity between two observations.

```python
from scipy.spatial.distance import pdist, squareform

euclidean_distances = pdist(
    X_hierarchical,
    metric="euclidean"
)

distance_matrix = squareform(euclidean_distances)

distance_matrix[:5, :5]
```

### Observation

The Euclidean distance matrix shows the pairwise distances between the first five observations using the nine standardized clinical variables.

- The diagonal values are **0**, because each observation has zero distance from itself.
- The distance between **Observation 0 and Observation 3** is **0.556283**, indicating that these two observations are relatively similar in the standardized feature space.
- The distance between **Observation 1 and Observation 2** is **1.359005**, showing greater similarity than several other pairs.
- The distance between **Observation 0 and Observation 1** is **3.470349**, indicating a larger difference between these observations.
- Smaller Euclidean distances indicate greater similarity, while larger distances indicate greater dissimilarity.
- These pairwise distances form the basis for the subsequent hierarchical clustering process.

### 2.5.3 Linkage Methods

Hierarchical clustering progressively combines observations or clusters based on their distances.

Different linkage methods define the distance between two clusters differently.

The following linkage methods are examined:

- **Single Linkage:** uses the minimum distance between observations belonging to two clusters.
- **Complete Linkage:** uses the maximum distance between observations belonging to two clusters.
- **Average Linkage:** uses the average distance between observations belonging to two clusters.
- **Ward Linkage:** merges clusters while minimizing the increase in within-cluster variance.

Dendrograms are generated for each linkage method to visualize the hierarchical structure of the observations.

```python
from scipy.cluster.hierarchy import linkage, dendrogram

linkage_methods = {
    "Single": "single",
    "Complete": "complete",
    "Average": "average",
    "Ward": "ward"
}

linkage_results = {}

for name, method in linkage_methods.items():
    linkage_results[name] = linkage(
        X_hierarchical,
        method=method,
        metric="euclidean"
    )
```

### 2.5.4 Single Linkage Dendrogram

The single linkage dendrogram shows how observations and clusters are progressively merged using the minimum distance between two clusters.

The vertical axis represents the distance at which clusters are merged. A large increase in merge distance indicates that relatively dissimilar clusters are being combined.

Because the dataset contains a large number of observations, only the final stages of the hierarchy are displayed to keep the dendrogram readable.

```python
from scipy.cluster.hierarchy import dendrogram
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

dendrogram(
    linkage_results["Single"],
    truncate_mode="lastp",
    p=30,
    show_leaf_counts=True
)

plt.title("Single Linkage Dendrogram")
plt.xlabel("Cluster")
plt.ylabel("Euclidean Distance")
plt.show()
```

### Observation

The single linkage dendrogram shows the hierarchical merging of observations based on the minimum distance between clusters.

- Most clusters are merged at relatively small Euclidean distances, mainly below approximately **4**.
- A few final merges occur at much larger distances, particularly around **5–6** and approximately **9.3**.
- The large increase in distance near the final merge indicates that the remaining groups are considerably different from each other.
- Single linkage can produce a chaining effect, where observations are connected through a sequence of relatively close observations.
- The dendrogram therefore provides a visual representation of how the observations progressively form larger clusters.

### 2.5.5 Complete Linkage Dendrogram

The complete linkage method measures the distance between two clusters using the maximum distance between their observations.

Compared with single linkage, complete linkage tends to form more compact clusters because a cluster is merged only when its farthest observations are relatively close.

The dendrogram is used to examine the hierarchical structure and the distances at which clusters are merged.

```python
plt.figure(figsize=(14, 6))

dendrogram(
    linkage_results["Complete"],
    truncate_mode="lastp",
    p=30,
    show_leaf_counts=True
)

plt.title("Complete Linkage Dendrogram")
plt.xlabel("Cluster")
plt.ylabel("Euclidean Distance")
plt.show()
```

### Observation

The complete linkage dendrogram shows a more compact hierarchical structure compared with single linkage.

- Most cluster merges occur at Euclidean distances below approximately **13**.
- Several groups are merged gradually between approximately **5 and 12**.
- A noticeable increase occurs near the final merges, with distances around **16** and **22**.
- The final merge at approximately **22** indicates that the two remaining groups are substantially different according to the standardized clinical variables.
- Complete linkage considers the maximum distance between observations of two clusters, which results in more compact cluster formation than single linkage.

### 2.5.6 Average Linkage Dendrogram

The average linkage method measures the distance between two clusters using the average distance between all pairs of observations belonging to the two clusters.

It provides a balance between single linkage and complete linkage by considering the overall separation between clusters rather than only the closest or farthest pair of observations.

The dendrogram is used to examine the sequence of cluster formation and the distances at which major groups are merged.

```python
plt.figure(figsize=(14, 6))

dendrogram(
    linkage_results["Average"],
    truncate_mode="lastp",
    p=30,
    show_leaf_counts=True
)

plt.title("Average Linkage Dendrogram")
plt.xlabel("Cluster")
plt.ylabel("Euclidean Distance")
plt.show()
```

### Observation

The average linkage dendrogram shows the hierarchical merging of clusters based on the average distance between observations.

- Most clusters are merged at relatively smaller distances, mainly below approximately **7**.
- Several larger groups are formed between approximately **7 and 12**.
- A noticeable increase occurs near the final stages, with merges around **12.3** and approximately **19.5**.
- The final merge occurs at the highest distance, indicating substantial separation between the two remaining groups.
- Compared with single linkage, average linkage reduces the chaining effect by considering the average distance between clusters.

### 2.5.7 Ward Linkage Dendrogram

Ward linkage forms clusters by minimizing the increase in within-cluster variance when two clusters are merged.

Since the clinical variables have already been standardized and Euclidean distance is being used, Ward linkage can be applied to the standardized numerical dataset.

The dendrogram shows how observations progressively form larger groups and helps identify major divisions in the hierarchical structure.

```python
plt.figure(figsize=(14, 6))

dendrogram(
    linkage_results["Ward"],
    truncate_mode="lastp",
    p=30,
    show_leaf_counts=True
)

plt.title("Ward Linkage Dendrogram")
plt.xlabel("Cluster")
plt.ylabel("Euclidean Distance")
plt.show()
```

### Observation

The Ward linkage dendrogram shows a clear hierarchical structure based on the increase in within-cluster variance during merging.

- The early merges occur at relatively smaller distances, mostly below approximately **100**.
- Larger groups are progressively formed between approximately **100 and 155**.
- A substantial increase in merge distance occurs after approximately **155**, followed by merges around **195**, **207**, and **242**.
- The final merge occurs at approximately **242**, indicating a large separation between the two final groups.
- Ward linkage produces relatively compact and internally similar clusters because it minimizes the increase in within-cluster variance during each merge.
- The large jumps in the upper part of the dendrogram indicate meaningful hierarchical divisions that can be examined when selecting the final number of clusters.

### 2.5.8 Comparison of Linkage Methods

The four linkage methods produce different hierarchical structures because they use different definitions of distance between clusters.

| Linkage Method | Cluster Distance Based On | Main Characteristic |
|---|---|---|
| Single | Minimum pairwise distance | Can produce chaining |
| Complete | Maximum pairwise distance | Produces compact clusters |
| Average | Average pairwise distance | Balances minimum and maximum distances |
| Ward | Increase in within-cluster variance | Favors compact, homogeneous clusters |

The dendrograms provide different views of the same standardized clinical data. The linkage method determines how observations are progressively combined into larger clusters.
